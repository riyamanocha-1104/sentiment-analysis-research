```python
#BoW
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer 
from nltk.tokenize import word_tokenize

def bag_of_words(df):
#     bow_vectorizer = CountVectorizer(max_df=0.90, min_df=0.2, stop_words=None, tokenizer=word_tokenize) 
    bow_vectorizer = CountVectorizer() 
    bow = bow_vectorizer.fit_transform(df['lemma_sentence(with POS)']) 
    #print(bow_vectorizer.get_feature_names())
    #print(bow_vectorizer.vocabulary_)
    return bow

df_bow=bag_of_words(df)
df_bow.shape
```

```python
#TF-IDF
def tf_idf(df):
#     tf_idf_vectorizer = TfidfVectorizer(max_df=0.9, min_df=0.2, stop_words=None, tokenizer=word_tokenize, norm='l2') 
    tf_idf_vectorizer = TfidfVectorizer(norm='l2') #extract features
    tfidf = tf_idf_vectorizer.fit_transform(df['lemma_sentence(with POS)']) #vectors
    return tfidf
df_tfidf=tf_idf(df)
df_tfidf.shape
```

```python
#Word2vec
#reference：https://www.pythonf.cn/read/93491
#https://github.com/Shwetago/Sentiment_Analysis/blob/master/Twitter_Sentiment_Analysis.ipynb
from nltk.tokenize import word_tokenize
Tokenize_tweet = df['lemma_sentence(with POS)'].apply(word_tokenize)
print(Tokenize_tweet)

Model_W2V = gensim.models.Word2Vec(Tokenize_tweet, size=200, #features
                                   window=5, 
                                   min_count=1, 
                                   sg=1,  #skip-gram model
                                   hs=0,
                                   negative=10, 
                                   workers=2, 
                                   seed=34 ) 
```

```python
#Each word can get its own vector. The representation of a tweets can the vector sum of each word divided by the total number(average) 
#or just the sum of each word vector
def word2vec_tweet(tokens, size):
    vector=np.zeros(size).reshape((1,size))
    vector_cnt = 0
    for word in tokens:
        vector += Model_W2V[word].reshape((1, size))
        vector_cnt += 1
    return vector/vector_cnt  #average for tweets

def word2vec_tweet_2(tokens, size):
    vector=np.zeros(size).reshape((1,size))
    vector_cnt = 0
    for word in tokens:
        vector += Model_W2V[word].reshape((1, size))
    return vector  #sum of tweets

tweet_arr=np.zeros((len(Tokenize_tweet), 200))

for i in range (len(Tokenize_tweet)):
    tweet_arr[i,:] = word2vec_tweet(Tokenize_tweet[i], 200)
tweet_vec_df = pd.DataFrame(tweet_arr)
tweet_vec_df
```

```python
# Create sentiment labels using VADER

from nltk.sentiment import SentimentIntensityAnalyzer

sia = SentimentIntensityAnalyzer()

def get_sentiment(text):
    score = sia.polarity_scores(text)["compound"]
    
    if score >= 0.05:
        return 1       # Positive
    elif score <= -0.05:
        return -1      # Negative
    else:
        return 0       # Neutral

df["sentiment"] = df["tweet"].apply(get_sentiment)

print(df[["tweet", "sentiment"]].head(10))
print("\nSentiment distribution:")
print(df["sentiment"].value_counts())
```

```python
# Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    df["tweet"],
    df["sentiment"],
    test_size=0.20,
    random_state=42,
    stratify=df["sentiment"]
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
# Convert tweets into TF-IDF features

tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2)
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)
```

```python
# Train Linear SVM model

svm_model = LinearSVC(random_state=42)

svm_model.fit(X_train_tfidf, y_train)

# Make predictions
y_pred = svm_model.predict(X_test_tfidf)

# Evaluate model
accuracy = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred, average="macro")

print("Linear SVM Results")
print("------------------")
print("Accuracy:", round(accuracy, 4))
print("Macro F1 Score:", round(f1, 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Negative", "Neutral", "Positive"]
))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from scipy.sparse import hstack

# Word-level TF-IDF
word_tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2)
)

# Character-level TF-IDF
char_tfidf = TfidfVectorizer(
    analyzer="char",
    max_features=5000,
    ngram_range=(3, 5)
)

X_train_word = word_tfidf.fit_transform(X_train)
X_test_word = word_tfidf.transform(X_test)

X_train_char = char_tfidf.fit_transform(X_train)
X_test_char = char_tfidf.transform(X_test)

# Combine word + character features
X_train_combined = hstack([X_train_word, X_train_char])
X_test_combined = hstack([X_test_word, X_test_char])

print("Combined training shape:", X_train_combined.shape)
print("Combined testing shape:", X_test_combined.shape)
```

```python
# Train Linear SVM using combined Word + Character TF-IDF features

combined_svm = LinearSVC(random_state=42)

combined_svm.fit(X_train_combined, y_train)

# Predictions
y_pred_combined = combined_svm.predict(X_test_combined)

# Evaluation
accuracy_combined = accuracy_score(y_test, y_pred_combined)
f1_combined = f1_score(y_test, y_pred_combined, average="macro")

print("Word + Character TF-IDF + Linear SVM Results")
print("---------------------------------------------")
print("Accuracy:", round(accuracy_combined, 4))
print("Macro F1 Score:", round(f1_combined, 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_combined,
    target_names=["Negative", "Neutral", "Positive"]
))
```

```python
# Hyperparameter tuning for Linear SVM

for c_value in [0.5, 1.0, 1.5, 2.0]:

    tuned_svm = LinearSVC(
        C=c_value,
        random_state=42
    )

    tuned_svm.fit(X_train_combined, y_train)

    tuned_pred = tuned_svm.predict(X_test_combined)

    tuned_accuracy = accuracy_score(y_test, tuned_pred)
    tuned_f1 = f1_score(y_test, tuned_pred, average="macro")

    print("C =", c_value)
    print("Accuracy:", round(tuned_accuracy, 4))
    print("Macro F1:", round(tuned_f1, 4))
    print("----------------------")
```

```python
# Final Improved Model
# Word + Character TF-IDF + Linear SVM (C=0.5)

final_svm = LinearSVC(
    C=0.5,
    random_state=42
)

final_svm.fit(X_train_combined, y_train)

final_pred = final_svm.predict(X_test_combined)

final_accuracy = accuracy_score(y_test, final_pred)
final_f1 = f1_score(y_test, final_pred, average="macro")

print("FINAL IMPROVED MODEL")
print("===================")
print("Accuracy:", round(final_accuracy, 4))
print("Macro F1 Score:", round(final_f1, 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    final_pred,
    target_names=["Negative", "Neutral", "Positive"]
))
```
